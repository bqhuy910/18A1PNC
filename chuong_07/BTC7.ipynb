```python
import tkinter as tk
from tkinter import filedialog, messagebox
from PIL import Image, ImageTk
import urllib.request
import io

cua_so = tk.Tk()
cua_so.title("Bài 7.5 - Xem ảnh")
cua_so.geometry("800x520")

nhan_anh = tk.Label(cua_so)
nhan_anh.pack(pady=10)

duong_dan_url = tk.StringVar()

anh_tk = None  # giữ tham chiếu để ảnh không bị mất

def hien_thi_anh_pil(anh_pil):
    global anh_tk
    anh_pil = anh_pil.resize((700, 380))
    anh_tk = ImageTk.PhotoImage(anh_pil)
    nhan_anh.config(image=anh_tk)

def mo_file_anh():
    tep = filedialog.askopenfilename(
        title="Chọn ảnh",
        filetypes=[("Image files", "*.png *.jpg *.jpeg *.gif *.bmp"), ("All files", "*.*")]
    )
    if not tep:
        return
    try:
        anh = Image.open(tep)
        hien_thi_anh_pil(anh)
    except Exception as e:
        messagebox.showerror("Lỗi", f"Không đọc được ảnh!\n{e}")

def tai_anh_tu_url():
    url = duong_dan_url.get().strip()
    if not url:
        messagebox.showwarning("Thiếu URL", "Nhập URL ảnh trước!")
        return
    try:
        du_lieu = urllib.request.urlopen(url).read()
        anh = Image.open(io.BytesIO(du_lieu))
        hien_thi_anh_pil(anh)
    except Exception as e:
        messagebox.showerror("Lỗi", f"Tải/đọc ảnh từ URL thất bại!\n{e}")

khung = tk.Frame(cua_so)
khung.pack()

tk.Button(khung, text="Mở ảnh từ máy", command=mo_file_anh).grid(row=0, column=0, padx=8)
tk.Entry(khung, textvariable=duong_dan_url, width=55).grid(row=0, column=1, padx=8)
tk.Button(khung, text="Tải ảnh từ URL", command=tai_anh_tu_url).grid(row=0, column=2, padx=8)

cua_so.mainloop()
```

```python
import tkinter as tk
from tkinter import ttk, messagebox

def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return abs(a)

def lcm(a, b):
    if a == 0 or b == 0:
        return 0
    return abs(a * b) // gcd(a, b)

cua_so = tk.Tk()
cua_so.title("Bài 7.9 - GCD/LCM")
cua_so.geometry("520x220")

tk.Label(cua_so, text="Enter value of m:", font=("Arial", 12)).grid(row=0, column=0, padx=10, pady=10, sticky="e")
tk.Label(cua_so, text="Enter value of n:", font=("Arial", 12)).grid(row=1, column=0, padx=10, pady=10, sticky="e")
tk.Label(cua_so, text="Choose function:", font=("Arial", 12)).grid(row=2, column=0, padx=10, pady=10, sticky="e")

o_m = tk.Entry(cua_so, width=20)
o_n = tk.Entry(cua_so, width=20)
o_m.grid(row=0, column=1)
o_n.grid(row=1, column=1)

lua_chon = tk.StringVar(value="GCD")
ttk.Combobox(cua_so, textvariable=lua_chon, values=["GCD", "LCM"], width=17, state="readonly").grid(row=2, column=1, sticky="w")

nhan_kq = tk.Label(cua_so, text="", font=("Arial", 12))
nhan_kq.grid(row=3, column=0, columnspan=2, pady=15)

def tinh():
    try:
        m = int(o_m.get())
        n = int(o_n.get())
    except:
        messagebox.showerror("Lỗi", "m và n phải là số nguyên!")
        return

    if lua_chon.get() == "GCD":
        kq = gcd(m, n)
        nhan_kq.config(text=f"GCD(m , n) = {kq}")
    else:
        kq = lcm(m, n)
        nhan_kq.config(text=f"LCM(m , n) = {kq}")

tk.Button(cua_so, text="Validate", command=tinh, width=14).grid(row=2, column=2, padx=10)
cua_so.mainloop()
```

```python
import tkinter as tk
from tkinter import messagebox

cua_so = tk.Tk()
cua_so.title("BMI Calculator")
cua_so.geometry("420x240")

tk.Label(cua_so, text="Cân nặng (kg):").pack(pady=8)
o_can = tk.Entry(cua_so, width=20)
o_can.pack()

tk.Label(cua_so, text="Chiều cao (m):").pack(pady=8)
o_cao = tk.Entry(cua_so, width=20)
o_cao.pack()

nhan_bmi = tk.Label(cua_so, text="BMI: ", font=("Arial", 12))
nhan_bmi.pack(pady=8)
nhan_ket_luan = tk.Label(cua_so, text="Kết luận: ", font=("Arial", 12, "bold"))
nhan_ket_luan.pack(pady=5)

def tinh_bmi():
    try:
        can = float(o_can.get())
        cao = float(o_cao.get())
        if can <= 0 or cao <= 0:
            raise ValueError
    except:
        messagebox.showerror("Lỗi", "Nhập cân nặng/chiều cao hợp lệ!")
        return

    bmi = can / (cao * cao)
    bmi = round(bmi, 2)
    nhan_bmi.config(text=f"BMI: {bmi}")

    if bmi < 18.5:
        kl = "Gầy"
    elif bmi < 25:
        kl = "Bình thường"
    elif bmi < 30:
        kl = "Thừa cân"
    else:
        kl = "Béo phì"
    nhan_ket_luan.config(text=f"Kết luận: {kl}")

tk.Button(cua_so, text="Tính BMI", command=tinh_bmi, width=14).pack(pady=10)
cua_so.mainloop()
```

```python
import tkinter as tk
from tkinter import ttk, messagebox

# Bảng giá theo ảnh bạn gửi
GIA_MO_CUA = 12000
KM_MO_CUA = 0.8

GIA_4CHO_TRONG_30 = 15300
GIA_4CHO_TU_31 = 12100

GIA_7CHO_TRONG_30 = 16100
GIA_7CHO_TU_31 = 13800

PHUT_MIEN_PHI = 5
GIA_CHO_MOI_PHUT = 750

cua_so = tk.Tk()
cua_so.title("Bài 7.14 - Tính cước taxi")
cua_so.geometry("520x300")

tk.Label(cua_so, text="Loại xe:").grid(row=0, column=0, padx=10, pady=10, sticky="e")
tk.Label(cua_so, text="Quãng đường (km):").grid(row=1, column=0, padx=10, pady=10, sticky="e")
tk.Label(cua_so, text="Thời gian chờ (phút):").grid(row=2, column=0, padx=10, pady=10, sticky="e")

loai_xe = tk.StringVar(value="4 chỗ")
ttk.Combobox(cua_so, textvariable=loai_xe, values=["4 chỗ", "7 chỗ"], state="readonly", width=18).grid(row=0, column=1, sticky="w")

o_km = tk.Entry(cua_so, width=22)
o_phut = tk.Entry(cua_so, width=22)
o_km.grid(row=1, column=1, sticky="w")
o_phut.grid(row=2, column=1, sticky="w")

nhan_kq = tk.Label(cua_so, text="Tổng tiền: ", font=("Arial", 13, "bold"))
nhan_kq.grid(row=4, column=0, columnspan=2, pady=20)

def tinh_tien_di(km, is_7cho):
    if km <= 0:
        return 0

    # tiền mở cửa
    tien = GIA_MO_CUA
    km_con_lai = km - KM_MO_CUA
    if km_con_lai <= 0:
        return tien

    if is_7cho:
        gia_trong_30 = GIA_7CHO_TRONG_30
        gia_tu_31 = GIA_7CHO_TU_31
    else:
        gia_trong_30 = GIA_4CHO_TRONG_30
        gia_tu_31 = GIA_4CHO_TU_31

    # Từ sau 0.8km đến 30km
    km_toi_da_trong_30 = max(0, 30 - KM_MO_CUA)
    km_trong_30 = min(km_con_lai, km_toi_da_trong_30)
    tien += km_trong_30 * gia_trong_30

    # Từ km 31 trở đi
    km_sau_30 = km_con_lai - km_trong_30
    if km_sau_30 > 0:
        tien += km_sau_30 * gia_tu_31

    return tien

def tinh_tien_cho(phut):
    if phut <= PHUT_MIEN_PHI:
        return 0
    return (phut - PHUT_MIEN_PHI) * GIA_CHO_MOI_PHUT

def tinh_tong():
    try:
        km = float(o_km.get())
        phut = int(float(o_phut.get()))
        if km <= 0 or phut < 0:
            raise ValueError
    except:
        messagebox.showerror("Lỗi", "Nhập km > 0 và phút chờ >= 0!")
        return

    is_7cho = (loai_xe.get() == "7 chỗ")
    tien_di = tinh_tien_di(km, is_7cho)
    tien_cho = tinh_tien_cho(phut)
    tong = int(round(tien_di + tien_cho))

    nhan_kq.config(text=f"Tổng tiền: {tong:,} đ")

tk.Button(cua_so, text="Tính tiền", command=tinh_tong, width=14).grid(row=3, column=1, pady=10, sticky="w")
cua_so.mainloop()
```
