```python
import sqlite3

def mo_ket_noi():
    kn = sqlite3.connect("product.db")
    kn.execute("""
    CREATE TABLE IF NOT EXISTS product (
        Id INTEGER PRIMARY KEY,
        Name TEXT NOT NULL,
        Price REAL NOT NULL,
        Amount INTEGER NOT NULL
    )
    """)
    kn.commit()
    return kn

def hien_thi(kn):
    cur = kn.cursor()
    cur.execute("SELECT * FROM product ORDER BY Id")
    ds = cur.fetchall()
    print("\n--- DANH SÁCH SẢN PHẨM ---")
    if not ds:
        print("(trống)")
    else:
        for d in ds:
            print(d)

def them(kn):
    Id = int(input("Nhập Id: "))
    Name = input("Nhập Name: ")
    Price = float(input("Nhập Price: "))
    Amount = int(input("Nhập Amount: "))
    kn.execute("INSERT INTO product (Id, Name, Price, Amount) VALUES (?,?,?,?)", (Id, Name, Price, Amount))
    kn.commit()
    print("Đã thêm!")

def tim_theo_ten(kn):
    ten = input("Nhập tên cần tìm: ")
    cur = kn.cursor()
    cur.execute("SELECT * FROM product WHERE Name LIKE ?", (f"%{ten}%",))
    ds = cur.fetchall()
    print("\n--- KẾT QUẢ TÌM ---")
    if not ds:
        print("(không thấy)")
    else:
        for d in ds:
            print(d)

def cap_nhat(kn):
    Id = int(input("Nhập Id cần cập nhật: "))
    Price = float(input("Nhập Price mới: "))
    Amount = int(input("Nhập Amount mới: "))
    cur = kn.cursor()
    cur.execute("UPDATE product SET Price=?, Amount=? WHERE Id=?", (Price, Amount, Id))
    kn.commit()
    print("Đã cập nhật!" if cur.rowcount else "Không thấy Id!")

def xoa(kn):
    Id = int(input("Nhập Id cần xóa: "))
    cur = kn.cursor()
    cur.execute("DELETE FROM product WHERE Id=?", (Id,))
    kn.commit()
    print("Đã xóa!" if cur.rowcount else "Không thấy Id!")

kn = mo_ket_noi()

while True:
    print("\n===== QUẢN LÝ SẢN PHẨM =====")
    print("1. Hiển thị danh sách")
    print("2. Thêm sản phẩm")
    print("3. Tìm theo tên")
    print("4. Cập nhật Price & Amount theo Id")
    print("5. Xóa theo Id")
    print("0. Thoát")
    chon = input("Chọn: ").strip()

    if chon == "1":
        hien_thi(kn)
    elif chon == "2":
        them(kn)
    elif chon == "3":
        tim_theo_ten(kn)
    elif chon == "4":
        cap_nhat(kn)
    elif chon == "5":
        xoa(kn)
    elif chon == "0":
        break
    else:
        print("Chọn sai!")

kn.close()
print("Đã thoát.")
```

```python
import sqlite3

def mo_ket_noi():
    kn = sqlite3.connect("ql_nhan_vien.db")
    kn.execute("PRAGMA foreign_keys = ON;")

    kn.execute("""
    CREATE TABLE IF NOT EXISTS PHONG (
        Id INTEGER PRIMARY KEY,
        Name TEXT NOT NULL
    )
    """)

    kn.execute("""
    CREATE TABLE IF NOT EXISTS NHAN_VIEN (
        id INTEGER PRIMARY KEY,
        ho_ten TEXT NOT NULL,
        tuoi INTEGER NOT NULL,
        dia_chi TEXT NOT NULL,
        luong REAL NOT NULL,
        Id_phong INTEGER,
        FOREIGN KEY (Id_phong) REFERENCES PHONG(Id)
    )
    """)
    kn.commit()
    return kn

def them_phong(kn):
    Id = int(input("Nhập Id phòng: "))
    Name = input("Nhập tên phòng: ")
    kn.execute("INSERT INTO PHONG (Id, Name) VALUES (?,?)", (Id, Name))
    kn.commit()
    print("Đã thêm phòng!")

def xem_phong(kn):
    cur = kn.cursor()
    cur.execute("SELECT * FROM PHONG ORDER BY Id")
    print("\n--- PHÒNG ---")
    for d in cur.fetchall():
        print(d)

def them_nhan_vien(kn):
    id_nv = int(input("Nhập id nhân viên: "))
    ho_ten = input("Nhập họ tên: ")
    tuoi = int(input("Nhập tuổi: "))
    dia_chi = input("Nhập địa chỉ: ")
    luong = float(input("Nhập lương: "))
    Id_phong = int(input("Nhập Id_phong: "))
    kn.execute("""
        INSERT INTO NHAN_VIEN (id, ho_ten, tuoi, dia_chi, luong, Id_phong)
        VALUES (?,?,?,?,?,?)
    """, (id_nv, ho_ten, tuoi, dia_chi, luong, Id_phong))
    kn.commit()
    print("Đã thêm nhân viên!")

def xem_nhan_vien(kn):
    cur = kn.cursor()
    cur.execute("""
        SELECT nv.id, nv.ho_ten, nv.tuoi, nv.dia_chi, nv.luong, p.Name
        FROM NHAN_VIEN nv LEFT JOIN PHONG p ON nv.Id_phong = p.Id
        ORDER BY nv.id
    """)
    print("\n--- NHÂN VIÊN (kèm phòng) ---")
    for d in cur.fetchall():
        print(d)

def tim_nhan_vien_theo_ten(kn):
    ten = input("Nhập tên cần tìm: ")
    cur = kn.cursor()
    cur.execute("SELECT * FROM NHAN_VIEN WHERE ho_ten LIKE ?", (f"%{ten}%",))
    ds = cur.fetchall()
    print("\n--- KẾT QUẢ TÌM ---")
    if not ds:
        print("(không thấy)")
    else:
        for d in ds:
            print(d)

def cap_nhat_luong_theo_id(kn):
    id_nv = int(input("Nhập id nhân viên cần cập nhật lương: "))
    luong_moi = float(input("Nhập lương mới: "))
    cur = kn.cursor()
    cur.execute("UPDATE NHAN_VIEN SET luong=? WHERE id=?", (luong_moi, id_nv))
    kn.commit()
    print("Đã cập nhật!" if cur.rowcount else "Không thấy id!")

def xoa_nhan_vien(kn):
    id_nv = int(input("Nhập id nhân viên cần xóa: "))
    cur = kn.cursor()
    cur.execute("DELETE FROM NHAN_VIEN WHERE id=?", (id_nv,))
    kn.commit()
    print("Đã xóa!" if cur.rowcount else "Không thấy id!")

kn = mo_ket_noi()

while True:
    print("\n===== QUẢN LÝ NHÂN VIÊN =====")
    print("1. Thêm phòng")
    print("2. Xem phòng")
    print("3. Thêm nhân viên")
    print("4. Xem nhân viên")
    print("5. Tìm nhân viên theo tên")
    print("6. Cập nhật lương theo id")
    print("7. Xóa nhân viên theo id")
    print("0. Thoát")
    chon = input("Chọn: ").strip()

    if chon == "1":
        them_phong(kn)
    elif chon == "2":
        xem_phong(kn)
    elif chon == "3":
        them_nhan_vien(kn)
    elif chon == "4":
        xem_nhan_vien(kn)
    elif chon == "5":
        tim_nhan_vien_theo_ten(kn)
    elif chon == "6":
        cap_nhat_luong_theo_id(kn)
    elif chon == "7":
        xoa_nhan_vien(kn)
    elif chon == "0":
        break
    else:
        print("Chọn sai!")

kn.close()
print("Đã thoát.")
```
