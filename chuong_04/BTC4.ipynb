```python
import pandas as pd

ser1 = pd.Series([1, 2, 3, 4, 5])
ser2 = pd.Series([4, 5, 6, 7, 8])

# 1) Từ ser1 xóa các mục có mặt trong ser2
ser1_sau_xoa = ser1[~ser1.isin(ser2)]

# 2) Lấy tất cả mục của ser1 và ser2 nhưng không nằm chung trong cả hai (symmetric difference)
muc_khong_chung = pd.Series(list(set(ser1).symmetric_difference(set(ser2))))

print("ser1_sau_xoa:\n", ser1_sau_xoa)
print("\nCác mục không chung:\n", muc_khong_chung.sort_values().reset_index(drop=True))
```

```python
import pandas as pd

euro12 = pd.read_csv("euro2012.csv")

print("type:", type(euro12))
print("shape:", euro12.shape)
print("columns:", euro12.columns.tolist())

# 1. In giá trị cột Goals
print("\n1) Goals:\n", euro12["Goals"])

# 2. Có bao nhiêu đội tham gia Euro2012?
print("\n2) Số đội tham gia:", euro12["Team"].nunique())

# 3. In thông tin Euro2012
print("\n3) info:")
euro12.info()

# 4. Tạo discipline gồm 3 cột
discipline = euro12[["Team", "Yellow Cards", "Red Cards"]]
print("\n4) discipline head:\n", discipline.head())

# 5. Sắp xếp discipline giảm dần theo Red Cards, Yellow Cards
discipline_sort = discipline.sort_values(by=["Red Cards", "Yellow Cards"], ascending=False)
print("\n5) discipline sort:\n", discipline_sort)

# 6a. Tính trung bình Yellow Cards
print("\n6a) Mean Yellow Cards:", discipline["Yellow Cards"].mean())

# 6b. Lọc đội ghi hơn 6 bàn
print("\n6b) Teams Goals > 6:\n", euro12.loc[euro12["Goals"] > 6, ["Team", "Goals"]])

# 7. Đội bắt đầu bằng 'G'
print("\n7) Teams startswith 'G':\n", euro12.loc[euro12["Team"].str.startswith("G"), ["Team"]])

# 8. In 7 cột đầu
print("\n8) 7 cột đầu:\n", euro12.iloc[:, :7].head())

# 9. In tất cả cột trừ 3 cột cuối
print("\n9) Tất cả cột trừ 3 cột cuối:\n", euro12.iloc[:, :-3].head())

# 10. In các cột Team, Goals, Shooting Accuracy, Yellow Cards, Red Cards
cot_can_in = ["Team", "Goals", "Shooting Accuracy", "Yellow Cards", "Red Cards"]
print("\n10) Các cột yêu cầu:\n", euro12[cot_can_in].head())

# 11. Chỉ hiện Team, Shooting Accuracy từ England, Italy, Russia
print("\n11) England/Italy/Russia:\n",
      euro12.loc[euro12["Team"].isin(["England", "Italy", "Russia"]), ["Team", "Shooting Accuracy"]])
```

```python
import pandas as pd

drink = pd.read_csv("drinks.csv", index_col=0)

# 1. type, shape, columns, head, tail
print("type:", type(drink))
print("shape:", drink.shape)
print("columns:", drink.columns.tolist())
print("\nhead:\n", drink.head())
print("\ntail:\n", drink.tail())

# 2. Số lượng bia tiêu thụ trung bình mỗi châu lục
print("\n2) Beer mean theo châu lục:\n", drink.groupby("continent")["beer_servings"].mean())

# 3. describe wine_servings theo châu lục
print("\n3) Describe wine_servings theo châu lục:\n", drink.groupby("continent")["wine_servings"].describe())

# 4. mean beer + wine theo châu lục
print("\n4) Mean beer & wine theo châu lục:\n",
      drink.groupby("continent")[["beer_servings", "wine_servings"]].mean())

# 5. median beer + wine theo châu lục
print("\n5) Median beer & wine theo châu lục:\n",
      drink.groupby("continent")[["beer_servings", "wine_servings"]].median())

# 6. spirit_servings mean/max/min theo châu lục
print("\n6) spirit_servings (mean/max/min) theo châu lục:\n",
      drink.groupby("continent")["spirit_servings"].agg(["mean", "max", "min"]))

# 7. sort_values theo beer_servings + top 5 nhiều nhất/ít nhất
drink_sort = drink.sort_values("beer_servings")
print("\n7) Top 5 uống bia ít nhất:\n", drink_sort.head(5)[["beer_servings", "continent"]])
print("\n7) Top 5 uống bia nhiều nhất:\n", drink_sort.tail(5)[["beer_servings", "continent"]])
```

```python
import numpy as np
import pandas as pd

stocks1 = pd.read_csv("stocks1.csv")
stocks2 = pd.read_csv("stocks2.csv")
companies = pd.read_csv("companies.csv")

print("=== stocks1 head/tail ===")
print(stocks1.head(5))
print(stocks1.tail(5))
print("\n=== stocks1 dtypes ===\n", stocks1.dtypes)
print("\n=== stocks1 info ===")
stocks1.info()

print("\n=== stocks2 head/tail ===")
print(stocks2.head(5))
print(stocks2.tail(5))
print("\n=== stocks2 dtypes ===\n", stocks2.dtypes)
print("\n=== stocks2 info ===")
stocks2.info()

print("\n=== companies xem dữ liệu ===")
print(companies.head())
print("\n=== companies dtypes ===\n", companies.dtypes)
print("\n=== companies info ===")
companies.info()

print("\n=== Kiểm tra Null stocks1 ===")
print(stocks1.isna().sum())

# thay Null theo quy tắc
stocks1["high"] = stocks1["high"].fillna(stocks1.groupby("symbol")["high"].transform("max"))
stocks1["low"]  = stocks1["low"].fillna(stocks1.groupby("symbol")["low"].transform("min"))

print("\n=== Sau khi thay Null ===")
print(stocks1.isna().sum())

# concat theo dòng
stocks = pd.concat([stocks1, stocks2], axis=0, ignore_index=True)
print("\n=== 15 dòng cuối của stocks ===")
print(stocks.tail(15))

# merge stocks với companies
stocks_companies = stocks.merge(companies, left_on="symbol", right_on="name", how="left")
print("\n=== 5 dòng đầu stocks_companies ===")
print(stocks_companies.head(5))

print("\n=== Mean open/high/low/close + volume theo công ty ===")
print(stocks_companies.groupby("symbol")[["open","high","low","close","volume"]].mean())

print("\n=== close: mean/max/min theo công ty ===")
print(stocks_companies.groupby("symbol")["close"].agg(["mean","max","min"]))

# parsed_time
stocks_companies["parsed_time"] = pd.to_datetime(stocks_companies["date"])
print("\n=== dtype parsed_time ===", stocks_companies["parsed_time"].dtype)
print("\n=== 5 dòng đầu sau khi có parsed_time ===")
print(stocks_companies.head(5))

# result
stocks_companies["result"] = np.where(stocks_companies["close"] > stocks_companies["open"], "up", "down")
print("\n=== 10 dòng đầu (date, symbol, open, close, result) ===")
print(stocks_companies[["date","symbol","open","close","result"]].head(10))
```

```python
import os, zipfile
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Nếu bạn có file zip: ml-latest-small.zip thì tự giải nén
tep_zip = "ml-latest-small.zip"
thu_muc = "ml-latest-small"

if os.path.exists(tep_zip) and (not os.path.exists(thu_muc)):
    with zipfile.ZipFile(tep_zip, "r") as z:
        z.extractall(".")

movies = pd.read_csv(f"{thu_muc}/movies.csv")
ratings = pd.read_csv(f"{thu_muc}/ratings.csv")
tags = pd.read_csv(f"{thu_muc}/tags.csv")

print("=== Cấu trúc dữ liệu ===")
print("movies:", movies.shape, "ratings:", ratings.shape, "tags:", tags.shape)

print("\n=== Thiếu dữ liệu (null) ===")
print("movies null:\n", movies.isna().sum())
print("ratings null:\n", ratings.isna().sum())
print("tags null:\n", tags.isna().sum())

# xử lý thiếu/không hợp lệ (cơ bản)
ratings = ratings.dropna(subset=["userId","movieId","rating"])

# gộp DataFrame
rm = ratings.merge(movies, on="movieId", how="left")

print("\n=== Thống kê rating ===")
print(ratings["rating"].describe())

# parsing timestamp
ratings["parsed_time"] = pd.to_datetime(ratings["timestamp"], unit="s")
tags["parsed_time"] = pd.to_datetime(tags["timestamp"], unit="s")

print("\n=== 5 dòng đầu ratings sau parse time ===")
print(ratings[["userId","movieId","rating","parsed_time"]].head())

# Lọc theo yêu cầu ví dụ: Top phim điểm cao (kèm điều kiện có nhiều lượt rating)
bang_tong = rm.groupby("title")["rating"].agg(["mean","count"]).sort_values(["mean","count"], ascending=False)
top10 = bang_tong[bang_tong["count"] >= 50].head(10)

print("\n=== Top 10 phim (mean rating cao, count>=50) ===")
print(top10)

# Trực quan hóa: histogram rating
plt.figure()
ratings["rating"].hist(bins=10)
plt.title("Phân bố rating")
plt.xlabel("rating")
plt.ylabel("tần suất")
plt.show()
```
