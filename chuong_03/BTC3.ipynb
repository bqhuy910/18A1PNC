## BÀi 3

```python
import numpy as np

arr_a = np.array([1,2,3,2,3,4,3,4,5,6])
arr_b = np.array([7,2,10,2,7,4,9,4,9,8])

# 1. Phần tử xuất hiện ở cả arr_a và arr_b (Giao nhau)
arr_c = np.intersect1d(arr_a, arr_b)
print("Phần tử chung:", arr_c)

# 2. Phần tử chỉ xuất hiện ở arr_a (Hiệu của 2 mảng)
arr_d = np.setdiff1d(arr_a, arr_b)
print("Chỉ có ở arr_a:", arr_d)

# 3. Lọc phần tử từ 5 đến 10
arr_e = np.array([2, 6, 1, 9, 10, 3, 27, 8, 6, 25, 16])
arr_f = arr_e[(arr_e >= 5) & (arr_e <= 10)]
print("Phần tử từ 5 đến 10:", arr_f)
```

```python
import pandas as pd
import numpy as np

#BƯỚC 1: TỰ TẠO DỮ LIỆU MẪU ĐỂ CHẠY 
data = {
    'country': ['Vietnam', 'Thailand', 'France', 'Germany', 'USA', 'Canada', 'Nigeria', 'South Africa'],
    'beer_servings': [163, 99, 127, 346, 249, 240, 31, 225],
    'spirit_servings': [70, 200, 151, 117, 158, 122, 2, 76],
    'wine_servings': [1, 1, 370, 175, 84, 100, 2, 81],
    'total_litres_of_pure_alcohol': [4.0, 6.4, 11.8, 11.3, 8.7, 8.2, 9.1, 8.2],
    'continent': ['AS', 'AS', 'EU', 'EU', 'NA', 'NA', 'AF', 'AF']
}
df_sample = pd.DataFrame(data)
df_sample.to_csv('drinks.csv', index=False)

#THỰC HIỆN CÁC YÊU CẦU BÀI 3.10 ---

# 1. Đọc dữ liệu, đặt cột đầu tiên (index_col=0) làm chỉ số
# Lưu ý: Vì file vừa tạo có cột đầu là 'country', nó sẽ thành index
drinks = pd.read_csv('drinks.csv', index_col=0)

# 2. Xem kiểu dữ liệu, kích thước, tên các cột
print("--- 2. Thông tin chung ---")
print(drinks.dtypes)
print(drinks.shape)
print(drinks.columns)

# 3. Xem 5 dòng đầu và cuối
print("\n--- 3. 5 dòng đầu ---")
print(drinks.head(5))
print("\n--- 3. 5 dòng cuối ---")
print(drinks.tail(5))

# 4. Số lượng bia tiêu thụ trung bình ở mỗi châu lục
print("\n--- 4. Bia TB mỗi châu lục ---")
print(drinks.groupby('continent')['beer_servings'].mean())

# 5. Thông tin thống kê về lượng rượu (wine_servings) ở mỗi châu lục
print("\n--- 5. Thống kê Wine mỗi châu lục ---")
print(drinks.groupby('continent')['wine_servings'].describe())

# 6. Mức tiêu thụ đồ uống trung bình mỗi châu lục (tất cả cột số)
print("\n--- 6. TB đồ uống theo châu lục ---")
print(drinks.groupby('continent').mean(numeric_only=True))

# 7. Mức tiêu thụ đồ uống trung vị (median) mỗi châu lục
print("\n--- 7. Trung vị theo châu lục ---")
print(drinks.groupby('continent').median(numeric_only=True))

# 8. Tiêu thụ rượu (spirit) Max, Min, Mean mỗi châu lục
print("\n--- 8. Thống kê Spirit (Max, Min, Mean) ---")
print(drinks.groupby('continent')['spirit_servings'].agg(['max', 'min', 'mean']))
```

## Bài 11

```python
import pandas as pd
import numpy as np

stocks1 = pd.DataFrame({
    "date":["2024-01-01","2024-01-02","2024-01-03"],
    "symbol":["AAA","AAA","AAA"],
    "open":[10,11,12],
    "high":[12,np.nan,13],
    "low":[9,10,np.nan],
    "close":[11,10,13],
    "volume":[1000,1100,1200]
})

stocks1["high"] = stocks1["high"].fillna(stocks1.groupby("symbol")["high"].transform("max"))
stocks1["low"] = stocks1["low"].fillna(stocks1.groupby("symbol")["low"].transform("min"))

stocks1["parsed_time"] = pd.to_datetime(stocks1["date"])
stocks1["result"] = np.where(stocks1["close"] > stocks1["open"], "up", "down")

print(stocks1)
```

```python
import pandas as pd

movies = pd.DataFrame({
    "movieId":[1,2,3],
    "title":["Toy Story","Jumanji","Heat"]
})

ratings = pd.DataFrame({
    "movieId":[1,1,2,3],
    "rating":[4.0,4.5,3.5,5.0],
    "timestamp":[964982703,964982224,964981247,964982400]
})

ratings["parsed_time"] = pd.to_datetime(ratings["timestamp"], unit="s")

rm = ratings.merge(movies, on="movieId")
print(rm.groupby("title")["rating"].mean())
```
